```python
"""
Small hands-on experiment with the COSMO lab's open-source PET-MAD model.

What it does (about 2 minutes on a laptop CPU, plus a one-time model download):
  1. Energy and forces of perfect diamond silicon (forces should be ~0 by symmetry).
  2. Rattle the atoms a little and let the model relax them back (geometry optimization).
  3. Scan the lattice constant and find the one the MODEL thinks is lowest in energy,
     then compare with the experimental value for silicon (about 5.43 angstrom).

Install first:   pip install upet ase numpy matplotlib
Source of the API used here: https://github.com/lab-cosmo/upet  (README)

Write down what you observe. Your notes are the real deliverable.
"""

import numpy as np
from ase.build import bulk
from ase.optimize import BFGS
from upet.ase import UPETCalculator

# Pinning the version makes your result reproducible.
calc = UPETCalculator(model="pet-mad-s", version="1.6.0", device="cpu")

A_EXPERIMENT = 5.431  # angstrom, silicon lattice constant (experiment)

# ---- 1) energy and forces of a perfect crystal --------------------------------
atoms = bulk("Si", "diamond", a=A_EXPERIMENT, cubic=True)
atoms.calc = calc
e = atoms.get_potential_energy()
f = atoms.get_forces()
print(f"[1] {len(atoms)} atoms, energy = {e:.4f} eV, max |force| = {np.abs(f).max():.2e} eV/A")

# ---- 2) rattle and relax ------------------------------------------------------
atoms.rattle(0.05, seed=42)  # random displacements of ~0.05 angstrom
e_before = atoms.get_potential_energy()
fmax_before = np.abs(atoms.get_forces()).max()
opt = BFGS(atoms, logfile=None)
opt.run(fmax=0.01)
e_after = atoms.get_potential_energy()
print(f"[2] rattled: E = {e_before:.4f} eV, max|F| = {fmax_before:.3f} eV/A")
print(f"    relaxed ({opt.get_number_of_steps()} steps): E = {e_after:.4f} eV, "
      f"max|F| = {np.abs(atoms.get_forces()).max():.3f} eV/A")

# ---- 3) lattice-constant scan -------------------------------------------------
a_values = np.linspace(5.20, 5.70, 11)
energies = []
for a in a_values:
    s = bulk("Si", "diamond", a=a, cubic=True)
    s.calc = calc
    energies.append(s.get_potential_energy() / len(s))  # eV per atom
energies = np.array(energies)

# fit a parabola near the minimum to locate it smoothly
c2, c1, c0 = np.polyfit(a_values, energies, 2)
a_model = -c1 / (2 * c2)
print(f"[3] model's preferred lattice constant ~ {a_model:.3f} A "
      f"(experiment {A_EXPERIMENT} A, difference {100 * (a_model - A_EXPERIMENT) / A_EXPERIMENT:+.2f} %)")

try:
    import matplotlib.pyplot as plt
    plt.plot(a_values, energies, "o-", label="PET-MAD")
    plt.axvline(A_EXPERIMENT, color="gray", ls="--", label="experiment")
    plt.xlabel("lattice constant (angstrom)")
    plt.ylabel("energy per atom (eV)")
    plt.legend()
    plt.savefig("si_lattice_scan.png", dpi=150, bbox_inches="tight")
    print("saved si_lattice_scan.png")
except ImportError:
    pass
```
